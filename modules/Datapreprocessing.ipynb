```python
import json
from tokenizers import Tokenizer

# Assuming your Args class and Tokenizer class are defined as before
class Args:
    def __init__(self):
        self.ann_path = '/workspace/MedGen/data/iu_xray/annotation.json'  # Path to your JSON annotations
        self.threshold = 3                           # Token frequency threshold
        self.dataset_name = 'iu_xray'               # Dataset name

# Initialize arguments and tokenizer
args = Args()
tokenizer = Tokenizer(args)

# Load the annotation data from JSON
with open(args.ann_path, 'r') as f:
    annotations = json.load(f)

# Process each report in the training set
for example in annotations['train']:
    report_id = example['id']
    report_text = example['report']
    
    print(f"Processing Report ID: {report_id}")
    print("Original Report:", report_text)
    
    # Clean and tokenize the report
    cleaned_report = tokenizer.clean_report_iu_xray(report_text)
    print("Cleaned Report:", cleaned_report)

    # Tokenize the cleaned report
    token_ids = tokenizer(cleaned_report)
    print("Token IDs:", token_ids)

    # Check if all token IDs are padding
    if all(id == 0 for id in token_ids):
        print("All token IDs are padding tokens.")
    
    # Decode back to text to verify correctness
    decoded_report = tokenizer.decode(token_ids)
    print("Decoded Report:", decoded_report)  # Should show the decoded text
    
    print("-" * 50)  # Separator for clarity
```
